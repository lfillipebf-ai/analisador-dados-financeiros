# Análise Financeira com Python e Pandas

Este notebook explora transações financeiras fictícias usando Pandas, SQL e visualizações.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/transacoes.csv', parse_dates=['data'])
df['valor'] = pd.to_numeric(df['valor'])
df['mes'] = df['data'].dt.to_period('M').astype(str)
df.head()
```

## Visão geral


```python
print('Registros:', len(df))
print('Período:', df['data'].min().date(), 'até', df['data'].max().date())
df.groupby('tipo')['valor'].agg(['count', 'sum']).round(2)
```

## Despesas por categoria


```python
despesas_categoria = (df[df['tipo'] == 'Despesa'].groupby('categoria')['valor'].sum().sort_values(ascending=False))
despesas_categoria.round(2)
```

```python
despesas_categoria.plot(kind='bar', title='Despesas por categoria')
plt.ylabel('Valor (R$)')
plt.xlabel('Categoria')
plt.tight_layout()
plt.show()
```

## Evolução mensal


```python
mensal = df.groupby(['mes', 'tipo'])['valor'].sum().unstack(fill_value=0)
mensal.round(2)
```

```python
mensal.plot(kind='bar', title='Receitas x Despesas por mês')
plt.ylabel('Valor (R$)')
plt.xlabel('Mês')
plt.tight_layout()
plt.show()
```

## Indicadores


```python
receitas = df.loc[df['tipo'] == 'Receita', 'valor'].sum()
despesas = df.loc[df['tipo'] == 'Despesa', 'valor'].sum()
saldo = receitas - despesas
taxa_poupanca = saldo / receitas * 100 if receitas else 0

print(f'Receitas: R$ {receitas:,.2f}')
print(f'Despesas: R$ {despesas:,.2f}')
print(f'Saldo: R$ {saldo:,.2f}')
print(f'Taxa de poupança: {taxa_poupanca:.2f}%')
```
