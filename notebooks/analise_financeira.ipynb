```python
import pandas as pd
df = pd.read_csv('../data/transacoes.csv', parse_dates=['data'])
df.head()
```

```python
df.groupby('tipo')['valor'].sum()
```

```python
df[df['tipo']=='Despesa'].groupby('categoria')['valor'].sum().sort_values(ascending=False)
```

```python
df['mes'] = df['data'].dt.to_period('M').astype(str)
df.groupby(['mes','tipo'])['valor'].sum().unstack(fill_value=0)
```
